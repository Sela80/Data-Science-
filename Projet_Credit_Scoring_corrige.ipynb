# 22 — Modélisation reproductible du risque de défaut

## Objectif métier
Étudier si les caractéristiques disponibles dans le dataset permettent d'estimer le risque qu'un prêt soit associé à un défaut (`loan_status = 1`). Le score est un outil d'aide à l'analyse, pas une décision automatique d'octroi ou de refus.

## Données et limites
Le notebook charge un dataset public référencé par KaggleHub (`nicholas33/credit-risk-dataset-csv`). Il ne s'agit pas de données identifiées comme provenant d'une banque ivoirienne. Les résultats ne sont donc pas directement généralisables à la Côte d'Ivoire ou à l'UMOA. Avant une utilisation réelle, il faudrait vérifier la fiche de source, la période, les définitions et les droits d'utilisation.

## Méthode
1. Vérifier les colonnes et la cible.
2. Créer un jeu de test final, puis une validation issue du train.
3. Traiter les anomalies déterministes et les valeurs manquantes sans apprendre de statistiques sur le test.
4. Comparer une régression logistique et une Random Forest sur la validation.
5. Choisir un modèle et un seuil à partir de la validation, puis évaluer une seule fois sur le test final.
6. Sauvegarder le pipeline complet pour reproduire les mêmes transformations en prédiction.

Les résultats chiffrés n'apparaîtront qu'après exécution des cellules dans Colab.

```python
# Imports dédiés à la modélisation
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, roc_auc_score, average_precision_score,
    precision_score, recall_score, f1_score,
    confusion_matrix, classification_report
)

RANDOM_STATE = 42
print("Imports de modélisation prêts.")
```

## 22.1 — Construire le jeu de modélisation

Les variables explicatives sont choisies pour représenter des informations susceptibles d'être disponibles lors de l'analyse d'une demande. Les variables `loan_grade` et `loan_int_rate` sont volontairement exclues à ce stade : leur disponibilité exacte au moment de la décision doit être vérifiée, et elles peuvent intégrer des décisions ou informations de risque déjà calculées.

Les valeurs extrêmes manifestement incohérentes (âge supérieur à 100 ans, ancienneté professionnelle supérieure à 60 ans) sont converties en valeurs manquantes. Ces règles sont des hypothèses de nettoyage à valider avec la documentation du dataset, et non des règles bancaires universelles.

```python
numeric_features = [
    "person_age",
    "person_income",
    "person_emp_length",
    "loan_amnt",
    "loan_percent_income",
    "cb_person_cred_hist_length"
]
categorical_features = [
    "person_home_ownership",
    "loan_intent",
    "cb_person_default_on_file"
]
features = numeric_features + categorical_features
target = "loan_status"

# Utiliser une copie de travail ; ne pas modifier df ni df_clean.
source_df = df_clean.copy() if "df_clean" in globals() else df.copy()

missing_columns = [c for c in features + [target] if c not in source_df.columns]
if missing_columns:
    raise ValueError(f"Colonnes absentes du dataset : {missing_columns}")

df_model = source_df[features + [target]].copy()
df_model = df_model.dropna(subset=[target]).copy()
df_model[target] = pd.to_numeric(df_model[target], errors="raise").astype(int)

target_values = set(df_model[target].unique())
if not target_values.issubset({0, 1}) or len(target_values) != 2:
    raise ValueError(f"La cible doit être codée en 0/1 avec les deux classes. Valeurs observées : {target_values}")

# Règles simples, documentées et déterministes ; les imputations apprises
# seront ajustées uniquement à partir des données d'entraînement.
df_model.loc[df_model["person_age"] > 100, "person_age"] = np.nan
df_model.loc[df_model["person_emp_length"] > 60, "person_emp_length"] = np.nan

X = df_model[features].copy()
y = df_model[target].copy()

print("Dimensions après préparation :", df_model.shape)
print("\nRépartition de la cible :")
display(pd.DataFrame({
    "Effectif": y.value_counts().sort_index(),
    "Pourcentage": (y.value_counts(normalize=True).sort_index() * 100).round(2)
}).rename(index={0: "Sans défaut (0)", 1: "Défaut (1)"}))
print(f"\nTaux de défaut observé : {y.mean():.2%}")
print("\nValeurs manquantes par variable :")
display(X.isna().sum().to_frame("Valeurs manquantes"))
```

## 22.2 — Séparer train, validation et test

Le jeu de test est isolé avant la comparaison des modèles. La validation sert à comparer les modèles et à sélectionner un seuil. Le test final ne doit pas servir à choisir le modèle ou le seuil : il donne une estimation plus honnête de la performance sur des observations non utilisées pendant la sélection.

```python
X_train_complet, X_test, y_train_complet, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train_complet, y_train_complet,
    test_size=0.20, random_state=RANDOM_STATE, stratify=y_train_complet
)

print("Entraînement :", X_fit.shape)
print("Validation   :", X_val.shape)
print("Test final   :", X_test.shape)
print(f"Taux de défaut — entraînement : {y_fit.mean():.2%}")
print(f"Taux de défaut — validation   : {y_val.mean():.2%}")
print(f"Taux de défaut — test         : {y_test.mean():.2%}")
```

## 22.3 — Définir les transformations dans un pipeline

L'imputation et l'encodage sont intégrés au pipeline : les médianes, modalités et transformations sont apprises sur le jeu d'entraînement seulement. `handle_unknown='ignore'` permet de traiter une catégorie non rencontrée pendant l'entraînement.

La standardisation est utile pour la régression logistique. Elle n'est généralement pas nécessaire pour une Random Forest, mais ne nuit pas au fonctionnement des arbres dans ce pipeline.

```python
numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
preprocessor = ColumnTransformer(transformers=[
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

models = {
    "Régression logistique": Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", LogisticRegression(
            max_iter=3000, class_weight="balanced", random_state=RANDOM_STATE
        ))
    ]),
    "Random Forest": Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(
            n_estimators=200,
            max_depth=12,
            min_samples_leaf=10,
            class_weight="balanced",
            random_state=RANDOM_STATE,
            n_jobs=-1
        ))
    ])
}

print("Pipelines créés :", ", ".join(models.keys()))
```

## 22.4 — Comparer les modèles sur la validation

L'Average Precision (résumé de la courbe précision-rappel) est utile lorsque la classe défaut est minoritaire. La ROC-AUC mesure la capacité de classement. Ces métriques évaluent la discrimination, mais ne suffisent pas à déterminer si les probabilités sont bien calibrées ni si le modèle est rentable pour une banque.

```python
validation_rows = []
validation_probabilities = {}

for name, pipeline in models.items():
    pipeline.fit(X_fit, y_fit)
    probabilities = pipeline.predict_proba(X_val)[:, 1]
    validation_probabilities[name] = probabilities
    validation_rows.append({
        "Modèle": name,
        "ROC-AUC": roc_auc_score(y_val, probabilities),
        "Average Precision": average_precision_score(y_val, probabilities)
    })

validation_results = pd.DataFrame(validation_rows).sort_values(
    "Average Precision", ascending=False
).reset_index(drop=True)

display(validation_results.round(4))
best_model_name = validation_results.loc[0, "Modèle"]
print("Modèle retenu provisoirement selon l'Average Precision de validation :", best_model_name)
print("Cette sélection reste exploratoire et doit être discutée au regard du coût métier des erreurs.")
```

## 22.5 — Étudier le compromis entre précision et rappel

Le seuil par défaut de 0,50 n'est pas automatiquement le meilleur seuil pour le risque de crédit. Un seuil plus bas peut détecter davantage de défauts (rappel plus élevé), mais aussi signaler davantage de dossiers sans défaut (baisse possible de la précision).

Ici, le seuil est choisi en maximisant le F1-score sur la validation parmi quelques seuils candidats. C'est une règle pédagogique, pas une politique de crédit validée : en pratique, il faudrait intégrer les coûts des faux négatifs et faux positifs, la capacité de revue humaine, la calibration et les contraintes applicables.

```python
threshold_rows = []
val_probabilities_best = validation_probabilities[best_model_name]

for threshold in [0.20, 0.30, 0.40, 0.50, 0.60]:
    val_pred = (val_probabilities_best >= threshold).astype(int)
    threshold_rows.append({
        "Seuil": threshold,
        "Précision défaut": precision_score(y_val, val_pred, zero_division=0),
        "Rappel défaut": recall_score(y_val, val_pred, zero_division=0),
        "F1 défaut": f1_score(y_val, val_pred, zero_division=0),
        "Accuracy": accuracy_score(y_val, val_pred)
    })

threshold_results = pd.DataFrame(threshold_rows)
display(threshold_results.round(4))

best_row_index = threshold_results["F1 défaut"].idxmax()
seuil_final = float(threshold_results.loc[best_row_index, "Seuil"])
print(f"Seuil provisoirement retenu sur validation : {seuil_final:.2f}")
print("Ne pas interpréter ce seuil comme un seuil réglementaire ou bancaire validé.")
```

## 22.6 — Évaluation finale sur le test

Le modèle retenu est réentraîné sur l'ensemble train + validation, puis évalué sur le jeu de test resté isolé. Les probabilités du test ne servent pas à modifier le seuil après coup.

```python
final_model = models[best_model_name]
final_model.fit(X_train_complet, y_train_complet)

test_probabilities = final_model.predict_proba(X_test)[:, 1]
test_predictions = (test_probabilities >= seuil_final).astype(int)

final_metrics = pd.DataFrame([{
    "Modèle": best_model_name,
    "Seuil": seuil_final,
    "Accuracy": accuracy_score(y_test, test_predictions),
    "ROC-AUC": roc_auc_score(y_test, test_probabilities),
    "Average Precision": average_precision_score(y_test, test_probabilities),
    "Précision défaut": precision_score(y_test, test_predictions, zero_division=0),
    "Rappel défaut": recall_score(y_test, test_predictions, zero_division=0),
    "F1 défaut": f1_score(y_test, test_predictions, zero_division=0)
}])

display(final_metrics.round(4))
print("\nMatrice de confusion — lignes = vraie classe, colonnes = classe prédite :")
cm = confusion_matrix(y_test, test_predictions, labels=[0, 1])
display(pd.DataFrame(
    cm,
    index=["Réel sans défaut (0)", "Réel défaut (1)"],
    columns=["Prédit sans défaut (0)", "Prédit défaut (1)"]
))

print("\nRapport de classification :")
print(classification_report(
    y_test, test_predictions,
    labels=[0, 1],
    target_names=["Sans défaut (0)", "Défaut (1)"],
    digits=4, zero_division=0
))

plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Prédit 0", "Prédit 1"],
            yticklabels=["Réel 0", "Réel 1"])
plt.title("Matrice de confusion — test final")
plt.xlabel("Classe prédite")
plt.ylabel("Classe réelle")
plt.tight_layout()
plt.show()
```

## 22.8 — Examiner l'importance des variables par permutation

L'importance par permutation mesure la baisse d'une métrique lorsqu'une variable est mélangée. Elle donne une indication de l'utilité prédictive dans ce jeu de test, mais ne prouve pas une causalité. Les variables corrélées peuvent se partager ou masquer leur importance.

```python
from sklearn.inspection import permutation_importance

permutation_result = permutation_importance(
    final_model,
    X_test,
    y_test,
    scoring="average_precision",
    n_repeats=5,
    random_state=RANDOM_STATE,
    n_jobs=-1
)
importance_df = pd.DataFrame({
    "Variable": X_test.columns,
    "Baisse moyenne de l'Average Precision": permutation_result.importances_mean,
    "Écart-type": permutation_result.importances_std
}).sort_values("Baisse moyenne de l'Average Precision", ascending=False)

display(importance_df.round(5))

top_importance = importance_df.head(10).sort_values(
    "Baisse moyenne de l'Average Precision", ascending=True
)
plt.figure(figsize=(9, 5))
plt.barh(top_importance["Variable"], top_importance["Baisse moyenne de l'Average Precision"])
plt.title("Importance par permutation — test final")
plt.xlabel("Baisse moyenne de l'Average Precision")
plt.tight_layout()
plt.show()
```
